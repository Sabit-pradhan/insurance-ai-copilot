```python
# Define project and data paths
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data" / "raw"

# Load underwriting data
underwriting = pd.read_csv(
    DATA_DIR / "underwriting.csv"
)

print("Dataset Shape:", underwriting.shape)

underwriting.head()
```

```python
# Numerical model features
numeric_features = [
    "AGE",
    "HEALTH_SCORE",
    "BMI",
    "CREDIT_SCORE"
]

# Categorical model features
categorical_features = [
    "LIFESTYLE",
    "MEDICAL_HISTORY_FLAG",
    "SMOKER_FLAG",
    "OCCUPATION_RISK"
]

# Combine both
all_features = numeric_features + categorical_features
```

```python
# Check whether every selected feature exists
missing_features = [
    col for col in all_features
    if col not in underwriting.columns
]

print("Missing features:", missing_features)
```

```python
# X = input features
X = underwriting[all_features].copy()

# y = target
y = underwriting["DECISION"].copy()

print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
# Import train-test split function
from sklearn.model_selection import train_test_split

# Split data into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train Shape:", X_train.shape)
print("Test Shape :", X_test.shape)
```

```python
# Tools for preprocessing numeric and categorical columns
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
```

```python
# Numeric preprocessing:
# 1. Fill missing values with median
# 2. Standardize numeric values

numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)
```

```python
# Categorical preprocessing:
# 1. Fill missing values with most frequent category
# 2. Convert categories into 0/1 columns

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
# Apply different preprocessing to different feature types
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
# Import Logistic Regression classifier
from sklearn.linear_model import LogisticRegression

# Create complete preprocessing + model pipeline
uw_lr = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        ("model", LogisticRegression(
            class_weight="balanced",
            max_iter=1500,
            random_state=42
        ))
    ]
)
```

```python
# Train underwriting Logistic Regression model
uw_lr.fit(
    X_train,
    y_train
)

print("Logistic Regression trained successfully")
```

```python
# Predict final underwriting class
lr_pred = uw_lr.predict(X_test)

# Predict probability for every class
lr_prob = uw_lr.predict_proba(X_test)
```

```python
# Import evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Calculate overall metrics
lr_accuracy = accuracy_score(
    y_test,
    lr_pred
)

lr_macro_f1 = f1_score(
    y_test,
    lr_pred,
    average="macro"
)

lr_weighted_f1 = f1_score(
    y_test,
    lr_pred,
    average="weighted"
)

print("Accuracy    :", round(lr_accuracy, 4))
print("Macro F1    :", round(lr_macro_f1, 4))
print("Weighted F1 :", round(lr_weighted_f1, 4))
```

```python
# Check precision, recall and F1 for each underwriting class
print(
    classification_report(
        y_test,
        lr_pred
    )
)
```

```python
# Check which classes model is confusing
lr_cm = confusion_matrix(
    y_test,
    lr_pred
)

print(lr_cm)
```

```python
# Import Random Forest classifier
from sklearn.ensemble import RandomForestClassifier

# Build preprocessing + Random Forest pipeline
uw_rf = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        ("model", RandomForestClassifier(
            n_estimators=300,        # Total trees
            max_depth=12,            # Maximum depth of each tree
            min_samples_split=10,    # Minimum rows needed to split
            min_samples_leaf=5,      # Minimum rows in a leaf
            class_weight="balanced", # Handle class imbalance
            random_state=42,
            n_jobs=-1
        ))
    ]
)
```

```python
# Train Random Forest
uw_rf.fit(
    X_train,
    y_train
)

print("Random Forest trained successfully")
```

```python
# Predict underwriting classes
rf_pred = uw_rf.predict(X_test)

# Predict class probabilities
rf_prob = uw_rf.predict_proba(X_test)
```

```python
# Calculate Random Forest metrics
rf_accuracy = accuracy_score(y_test, rf_pred)

rf_macro_f1 = f1_score(
    y_test,
    rf_pred,
    average="macro"
)

rf_weighted_f1 = f1_score(
    y_test,
    rf_pred,
    average="weighted"
)

print("Accuracy    :", round(rf_accuracy, 4))
print("Macro F1    :", round(rf_macro_f1, 4))
print("Weighted F1 :", round(rf_weighted_f1, 4))
```

```python
# Class-wise performance
print(
    classification_report(
        y_test,
        rf_pred
    )
)
```

```python
# Compare Logistic Regression and Random Forest
uw_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest"
    ],
    "Accuracy": [
        lr_accuracy,
        rf_accuracy
    ],
    "Macro F1": [
        lr_macro_f1,
        rf_macro_f1
    ],
    "Weighted F1": [
        lr_weighted_f1,
        rf_weighted_f1
    ]
})

uw_comparison
```

```python
# Convert text target classes into numeric labels
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

# See class-to-number mapping
class_mapping = dict(
    zip(
        label_encoder.classes_,
        label_encoder.transform(label_encoder.classes_)
    )
)

print(class_mapping)
```

```python
# Give more training importance to minority classes
from sklearn.utils.class_weight import compute_sample_weight

sample_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train_encoded
)

print("Training rows:", len(y_train_encoded))
print("Sample weights:", len(sample_weights))
```

```python
# Import XGBoost classifier
from xgboost import XGBClassifier

# Preprocessing + multiclass XGBoost pipeline
uw_xgb = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        ("model", XGBClassifier(
            objective="multi:softprob",   # Predict probability for each class
            num_class=len(label_encoder.classes_),

            n_estimators=300,             # Number of boosting trees
            max_depth=5,                  # Tree complexity
            learning_rate=0.05,           # Learning speed

            subsample=0.8,                # Use 80% rows per tree
            colsample_bytree=0.8,         # Use 80% features per tree

            eval_metric="mlogloss",
            random_state=42,
            n_jobs=-1
        ))
    ]
)
```

```python
# Train XGBoost with balanced sample weights
uw_xgb.fit(
    X_train,
    y_train_encoded,
    model__sample_weight=sample_weights
)

print("Underwriting XGBoost trained successfully")
```

```python
# Train XGBoost with balanced sample weights
uw_xgb.fit(
    X_train,
    y_train_encoded,
    model__sample_weight=sample_weights
)

print("Underwriting XGBoost trained successfully")
```

```python
# Predict numeric class labels
xgb_pred_encoded = uw_xgb.predict(X_test)

# Convert numeric predictions back to original class names
xgb_pred = label_encoder.inverse_transform(
    xgb_pred_encoded.astype(int)
)

# Probability for each underwriting class
xgb_prob = uw_xgb.predict_proba(X_test)
```

```python
# Calculate multiclass metrics
xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)

xgb_macro_f1 = f1_score(
    y_test,
    xgb_pred,
    average="macro"
)

xgb_weighted_f1 = f1_score(
    y_test,
    xgb_pred,
    average="weighted"
)

print("Accuracy    :", round(xgb_accuracy, 4))
print("Macro F1    :", round(xgb_macro_f1, 4))
print("Weighted F1 :", round(xgb_weighted_f1, 4))
```

```python
# Precision, Recall and F1 for every decision class
print(
    classification_report(
        y_test,
        xgb_pred
    )
)
```

```python
# Compare all underwriting models
uw_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],

    "Accuracy": [
        lr_accuracy,
        rf_accuracy,
        xgb_accuracy
    ],

    "Macro F1": [
        lr_macro_f1,
        rf_macro_f1,
        xgb_macro_f1
    ],

    "Weighted F1": [
        lr_weighted_f1,
        rf_weighted_f1,
        xgb_weighted_f1
    ]
})

uw_comparison
```
