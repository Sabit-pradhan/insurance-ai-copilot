```python
overview = []

for name, df in datasets.items():
    total_cells = df.shape[0] * df.shape[1]
    missing = df.isnull().sum().sum()

    overview.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": missing,
        "Missing %": round((missing / total_cells) * 100, 2),
        "Duplicate Rows": df.duplicated().sum()
    })

overview_df = pd.DataFrame(overview)

overview_df
```

```python
def inspect_dataset(df, name):
    print("=" * 70)
    print(f"DATASET: {name.upper()}")
    print("=" * 70)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nFirst 5 Rows:")
    display(df.head())
```

```python
# Foreign key validation

checks = {
    "Policy -> Customer": (~policies["CUSTOMER_ID"].isin(customers["CUSTOMER_ID"])).sum(),
    "Policy -> Agent": (~policies["AGENT_ID"].isin(agents["AGENT_ID"])).sum(),
    "Claim -> Policy": (~claims["POLICY_ID"].isin(policies["POLICY_ID"])).sum(),
    "Premium -> Policy": (~premiums["POLICY_ID"].isin(policies["POLICY_ID"])).sum(),
    "Renewal -> Policy": (~renewals["POLICY_ID"].isin(policies["POLICY_ID"])).sum(),
    "Underwriting -> Customer": (~underwriting["CUSTOMER_ID"].isin(customers["CUSTOMER_ID"])).sum()
}

pd.Series(checks)
```

```python
primary_keys = {
    "customers": "CUSTOMER_ID",
    "policies": "POLICY_ID",
    "claims": "CLAIM_ID",
    "premiums": "PREMIUM_TXN_ID",
    "renewals": "RENEWAL_ID",
    "underwriting": "UNDERWRITING_ID",
    "agents": "AGENT_ID"
}

for name, key in primary_keys.items():
    df = datasets[name]

    print(
        name,
        "| duplicates:", df[key].duplicated().sum(),
        "| missing:", df[key].isnull().sum()
    )
```
